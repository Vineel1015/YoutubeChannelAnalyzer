```python
from googleapiclient.discovery import build
import pandas as pd

API_KEY = "AIzaSyDppAtpVZwXITRCKy2Txd-KZxlNiBj8xi4"
youtube = build("youtube", "v3", developerKey=API_KEY)


def get_video_ids(channel_id, youtube):

  channel_data = []

  request1 = youtube.channels().list(
      id = channel_id,
      part = 'contentDetails'
  )

  response1 = request1.execute()
  upload_key = response1['items'][0]['contentDetails']['relatedPlaylists']['uploads']

  request2 = youtube.playlistItems().list(
      playlistId = upload_key,
      part = 'snippet',
      maxResults = 200
  )

  response2 = request2.execute()
  for i in range(0, 200):
    try:
      channel_data.append(response2['items'][i]['snippet']['resourceId']['videoId'])
    except:
      return channel_data

video_id_sam = get_video_ids('UC1DTYW241WD64ah5BFWn4JA',youtube)
video_id_sam
```

```python
def get_video_stats(video_id_list, youtube):

  video_data = []

  for item in video_id_list:
    request = youtube.videos().list(
        part = 'snippet,statistics',
        id = item
    )

    response = request.execute()
    video_data.append(
        {
            'channel_title': response['items'][0]['snippet']['channelTitle'],
            'category_id': response['items'][0]['snippet']['categoryId'],
            'view_count': response['items'][0]['statistics']['viewCount'],
            'like_count': response['items'][0]['statistics']['likeCount'],
            'comment_count': response['items'][0]['statistics']['commentCount']
        }
    )

  return video_data

sam_videos_data = get_video_stats(video_id_sam, youtube)
```

```python
import pandas as pd
category_id_list = []
view_count_list = []
like_count_list = []
comment_count_list = []


for item in sam_videos_data:
    category_id_list.append(item.get('category_id'))
    view_count_list.append(item.get('view_count'))
    like_count_list.append(item.get('like_count'))
    comment_count_list.append(item.get('comment_count'))


df_sam_videos = pd.DataFrame({
    'Category ID': category_id_list,
    'View Count': view_count_list,
    'Like Count': like_count_list,
    'Comment Count': comment_count_list
})


df_sam_videos
```
